# 50. Baseline 모델 빠르게 학습하기

가장 단순한 baseline segmentation 모델을 짧게 학습합니다. 처음부터 큰 모델을 쓰면 개선 효과를 판단하기 어렵기 때문에 작은 `TinyFCN`을 기준점으로 둡니다.

이번 노트북의 목표는 다음과 같습니다.

- 49장에서 만든 데이터셋을 불러옵니다.
- 작은 baseline 모델을 빠르게 학습합니다.
- `metrics.json`, `config.json`, `checkpoint.pt`, `pred_samples/`를 저장합니다.

```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / "seg7_utils.py").exists():
    NOTEBOOK_DIR = Path("Vision 기초/7장")

sys.path.append(str(NOTEBOOK_DIR))
DATA_ROOT = NOTEBOOK_DIR / "data" / "mini_shapes_seg"
RUNS_ROOT = NOTEBOOK_DIR / "runs"

from seg7_utils import *
set_korean_font()
set_seed(7)
```

## 50-1. 데이터셋 준비 확인

```python
if not (DATA_ROOT / "meta.json").exists():
    create_mini_shapes_dataset(DATA_ROOT)

inspect_dataset(DATA_ROOT)
```

## 50-2. TinyFCN baseline 학습

```python
model, metrics, history, run_dir = train_experiment(
    experiment_name="50_baseline_fcn",
    dataset_root=DATA_ROOT,
    runs_root=RUNS_ROOT,
    model_name="tiny_fcn",
    epochs=2,
    batch_size=16,
    lr=3e-3,
    loss_name="ce",
    augment=False,
    channels=8,
    seed=7,
)

metrics
```

## 50-3. 학습 곡선 확인

```python
import matplotlib.pyplot as plt

xs = [row["epoch"] for row in history]
plt.figure(figsize=(6, 3))
plt.plot(xs, [row["train_loss"] for row in history], marker="o", label="train loss")
plt.plot(xs, [row["val_loss"] for row in history], marker="o", label="val loss")
plt.xlabel("epoch")
plt.ylabel("loss")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"run_dir: {run_dir}")
```
