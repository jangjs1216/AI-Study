# 53. Loss와 Class Imbalance 개선 실험

baseline의 Cross Entropy loss를 `CE + Dice` 조합으로 바꿔 봅니다. Dice loss는 객체 영역의 겹침을 직접 압박하므로 class imbalance가 있는 segmentation 문제에서 비교할 가치가 있습니다.

```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / "seg7_utils.py").exists():
    NOTEBOOK_DIR = Path("Vision 기초/7장")

sys.path.append(str(NOTEBOOK_DIR))
DATA_ROOT = NOTEBOOK_DIR / "data" / "mini_shapes_seg"
RUNS_ROOT = NOTEBOOK_DIR / "runs"

from seg7_utils import *
set_korean_font()
set_seed(7)
```

## 53-1. CE + Dice loss 실험

```python
if not (DATA_ROOT / "meta.json").exists():
    create_mini_shapes_dataset(DATA_ROOT)

model, metrics, history, run_dir = train_experiment(
    experiment_name="53_loss_ce_dice",
    dataset_root=DATA_ROOT,
    runs_root=RUNS_ROOT,
    model_name="tiny_fcn",
    epochs=4,
    batch_size=16,
    lr=3e-3,
    loss_name="ce_dice",
    augment=False,
    channels=8,
    seed=11,
)

metrics
```

## 53-2. Baseline과 빠른 비교

```python
baseline = read_json(RUNS_ROOT / "50_baseline_fcn" / "metrics.json")
print("baseline mean IoU:", baseline["mean_iou"])
print("CE + Dice mean IoU:", metrics["mean_iou"])
print("저장 위치:", run_dir)
```
