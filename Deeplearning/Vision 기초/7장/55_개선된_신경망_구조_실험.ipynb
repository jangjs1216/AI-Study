# 55. 개선된 신경망 구조 실험

baseline `TinyFCN`보다 구조적으로 강한 `TinyUNet`을 학습합니다. U-Net은 encoder-decoder와 skip connection을 사용하므로 위치 정보와 문맥 정보를 함께 활용할 수 있습니다.

```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / "seg7_utils.py").exists():
    NOTEBOOK_DIR = Path("Vision 기초/7장")

sys.path.append(str(NOTEBOOK_DIR))
DATA_ROOT = NOTEBOOK_DIR / "data" / "mini_shapes_seg"
RUNS_ROOT = NOTEBOOK_DIR / "runs"

from seg7_utils import *
set_korean_font()
set_seed(7)
```

## 55-1. TinyUNet 개선 모델 학습

```python
if not (DATA_ROOT / "meta.json").exists():
    create_mini_shapes_dataset(DATA_ROOT)

model, metrics, history, run_dir = train_experiment(
    experiment_name="55_tiny_unet_improved",
    dataset_root=DATA_ROOT,
    runs_root=RUNS_ROOT,
    model_name="tiny_unet",
    epochs=6,
    batch_size=16,
    lr=2e-3,
    loss_name="ce_dice",
    augment=True,
    channels=12,
    seed=17,
)

metrics
```

## 55-2. 학습 history와 최종 산출물 확인

```python
import matplotlib.pyplot as plt

xs = [row["epoch"] for row in history]
plt.figure(figsize=(7, 3))
plt.plot(xs, [row["val_mean_iou"] for row in history], marker="o", label="val mIoU")
plt.plot(xs, [row["val_pixel_accuracy"] for row in history], marker="o", label="val pixel acc")
plt.ylim(0, 1)
plt.xlabel("epoch")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print("저장 위치:", run_dir)
print("sample predictions:", sorted((run_dir / "pred_samples").glob("*.png"))[:2])
```
