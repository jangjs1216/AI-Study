## 60-2. 오프라인 dense feature contract 확인


```python
from PIL import Image
import torch
import matplotlib.pyplot as plt

pairs = list_pairs(DATA_ROOT, "val")
image = Image.open(pairs[0][0]).convert("RGB")
image_np = np.asarray(image)
image_tensor = image_to_tensor(image_np).unsqueeze(0)

backbone = DenseFeatureAdapter("handcrafted")
with torch.no_grad():
    features = backbone(image_tensor)

print("image:", tuple(image_tensor.shape))
print("dense features:", tuple(features.shape))
```

## 60-3. 일부 feature channel 보기


```python
fig, axes = plt.subplots(1, 4, figsize=(10, 3))
for ax, idx in zip(axes, [0, 3, 6, 7]):
    ax.imshow(features[0, idx].numpy(), cmap="viridis")
    ax.set_title(f"feature {idx}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```
