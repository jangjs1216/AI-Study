## 재질별 texture 비교

표면 texture는 현실감을 주되, 모델이 texture id만 shortcut으로 외우지 않도록 variation을 둡니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(8)
materials = ["brushed_medium", "smooth_bright", "rough_dark"]
fig, axes = plt.subplots(1, len(materials), figsize=(9, 3))
for ax, material in zip(axes, materials):
    texture = make_surface_texture(rng, 128, material, roughness=0.45, base_color=(160, 165, 170))
    ax.imshow(np.clip(texture, 0, 255).astype("uint8"))
    ax.set_title(material)
    ax.axis("off")
plt.tight_layout()
```
