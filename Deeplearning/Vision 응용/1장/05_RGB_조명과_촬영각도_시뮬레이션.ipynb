# 05. RGB 조명과 촬영각도 시뮬레이션

RGB 조명과 촬영각도를 바꿔 같은 계열의 형상이 다른 appearance를 갖도록 만듭니다.

```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / "synthetic_metal_utils.py").exists():
    search_roots = [Path.cwd(), *Path.cwd().parents]
    search_patterns = ["synthetic_metal_utils.py", "*/synthetic_metal_utils.py", "*/*/synthetic_metal_utils.py"]
    for root in search_roots:
        for pattern in search_patterns:
            matches = list(root.glob(pattern))
            if matches:
                NOTEBOOK_DIR = matches[0].parent
                break
        if (NOTEBOOK_DIR / "synthetic_metal_utils.py").exists():
            break

sys.path.append(str(NOTEBOOK_DIR))
DATA_ROOT = NOTEBOOK_DIR / "data" / "synthetic_metal_seg"

from synthetic_metal_utils import *
set_korean_font()
set_seed(7)

print("NOTEBOOK_DIR:", NOTEBOOK_DIR)
print("DATA_ROOT:", DATA_ROOT)
```

## domain별 sample preview

light_rgb와 camera_angle이 바뀌면 같은 metal class라도 입력 분포가 달라집니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(9)
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for ax, spec in zip(axes.ravel(), DEFAULT_DOMAIN_SPECS[:10]):
    image, mask, meta = sample_from_domain(rng, spec, 128, spec.domain_id, 0)
    ax.imshow(make_overlay(image, mask))
    ax.set_title(spec.domain_id, fontsize=7)
    ax.axis("off")
plt.tight_layout()
```
