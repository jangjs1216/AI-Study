# 06. 메탈 반사 Highlight Glare 생성

금속 표면에서 자주 생기는 specular highlight와 glare를 의도적으로 생성합니다.

```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / "synthetic_metal_utils.py").exists():
    search_roots = [Path.cwd(), *Path.cwd().parents]
    search_patterns = ["synthetic_metal_utils.py", "*/synthetic_metal_utils.py", "*/*/synthetic_metal_utils.py"]
    for root in search_roots:
        for pattern in search_patterns:
            matches = list(root.glob(pattern))
            if matches:
                NOTEBOOK_DIR = matches[0].parent
                break
        if (NOTEBOOK_DIR / "synthetic_metal_utils.py").exists():
            break

sys.path.append(str(NOTEBOOK_DIR))
DATA_ROOT = NOTEBOOK_DIR / "data" / "synthetic_metal_seg"

from synthetic_metal_utils import *
set_korean_font()
set_seed(7)

print("NOTEBOOK_DIR:", NOTEBOOK_DIR)
print("DATA_ROOT:", DATA_ROOT)
```

## glare가 강한 sample

반사 highlight는 모델이 결함이나 경계로 오인할 수 있는 대표적인 shortcut입니다.

```python
import numpy as np
import matplotlib.pyplot as plt
from dataclasses import replace

rng = np.random.default_rng(10)
spec = replace(DEFAULT_DOMAIN_SPECS[-1], glare_prob=1.0)
fig, axes = plt.subplots(1, 6, figsize=(11, 2.2))
for idx, ax in enumerate(axes):
    image, mask, meta = sample_from_domain(rng, spec, 128, f"glare_{idx}", idx)
    ax.imshow(make_overlay(image, mask))
    ax.set_title(f"glare={meta['glare_strength']:.1f}", fontsize=8)
    ax.axis("off")
plt.tight_layout()
```
